## 3) OpenCV basics
Upload a car photo using the folder icon on the left and name it `car.jpg`.

```
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('car.jpg')
print('Shape (height, width, channels):', img.shape)
cv2_imshow(img)
```

### Crop
An image is a matrix: `img[y1:y2, x1:x2]`. Try different numbers until only the plate is left.

```
crop = img[200:260, 300:500]   # <-- change these numbers for your image
cv2_imshow(crop)
```

### Grayscale and 2x enlarge (used for OCR cleanup in Week 3)

```
gray = cv2.cvtColor(crop, cv2.COLOR_BGR2GRAY)
big = cv2.resize(gray, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
cv2_imshow(big)
cv2.imwrite(f'{BASE}/plates/test.jpg', big)
```

## 4) Helper for cropping the 50 plates
Give the coordinates for each photo; the crop is saved to Drive.

```
def crop_plate(path, x1, y1, x2, y2, name):
    img = cv2.imread(path)
    crop = img[y1:y2, x1:x2]
    out = f'{BASE}/plates/{name}'
    cv2.imwrite(out, crop)
    cv2_imshow(crop)
    print('Saved:', out)

# Example:
# crop_plate('car.jpg', 300, 200, 500, 260, 'plate_01.jpg')
```
